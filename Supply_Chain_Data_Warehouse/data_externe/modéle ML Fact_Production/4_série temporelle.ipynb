```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.preprocessing import StandardScaler
import pyodbc  # Si vous utilisez pyodbc pour la connexion SQL

# Connexion à la base de données SQL Server (ajustez les paramètres de connexion)
conn = pyodbc.connect('DRIVER={SQL Server};SERVER=LAPTOP-GPVK1ANC;DATABASE=DW_SupplyChain;UID=sa;PWD=0000')
# Récupérer les données de la table `fact_production`
query = """
SELECT Production_Start_Time, quantityused, Dosage, expirydate
FROM fact_production
WHERE Production_Start_Time IS NOT NULL
"""
df = pd.read_sql(query, conn)

# Fermer la connexion
conn.close()

# Convertir 'Production_Start_Time' en datetime pour pouvoir effectuer des opérations temporelles
df['Production_Start_Time'] = pd.to_datetime(df['Production_Start_Time'], format='%Y%m%d%H')

# 1. Ajouter des caractéristiques temporelles (jour de la semaine, mois, année, etc.)
df['day_of_week'] = df['Production_Start_Time'].dt.dayofweek
df['month'] = df['Production_Start_Time'].dt.month
df['year'] = df['Production_Start_Time'].dt.year

# 2. Ajouter des variables de lag (par exemple pour prédire 'quantityused' à partir des valeurs passées)
df['lag_1'] = df['quantityused'].shift(1)
df['lag_2'] = df['quantityused'].shift(2)
df['lag_3'] = df['quantityused'].shift(3)

# Calculer des moyennes mobiles
df['rolling_mean_3'] = df['quantityused'].rolling(window=3).mean()

# Supprimer les NaN générés par les décalages et moyennes mobiles
df = df.dropna()

# 3. Diviser les données en caractéristiques (X) et variable cible (y)
X = df[['lag_1', 'lag_2', 'lag_3', 'rolling_mean_3', 'day_of_week', 'month', 'year']]
y = df['quantityused']

# 4. Diviser les données en ensembles d'entraînement et de test (test_size=0.2 signifie 80% pour l'entraînement et 20% pour le test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# 5. Normaliser les données (important pour les modèles de régression linéaire)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 6. Entraîner un modèle de régression linéaire (vous pouvez aussi essayer d'autres modèles comme RandomForest, XGBoost)
model = LinearRegression()
model.fit(X_train_scaled, y_train)

# 7. Prédiction et évaluation du modèle
y_pred = model.predict(X_test_scaled)

# 8. Calcul des erreurs : MAE et MSE
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

# Affichage des résultats
print(f"MAE (Mean Absolute Error): {mae}")
print(f"MSE (Mean Squared Error): {mse}")

# Visualisation des prédictions vs réelles
plt.figure(figsize=(10, 6))
plt.plot(y_test.index, y_test, label='True Values', color='blue')
plt.plot(y_test.index, y_pred, label='Predictions', color='red')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Quantity Used')
plt.title('Predictions vs Actual Values')
plt.show()
```

```python
from sklearn.model_selection import TimeSeriesSplit

# Créer un objet TimeSeriesSplit pour la validation croisée temporelle
tscv = TimeSeriesSplit(n_splits=5)

mae_scores = []
mse_scores = []

for train_index, test_index in tscv.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    # Normalisation des données
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # Entraîner le modèle
    model = LinearRegression()
    model.fit(X_train_scaled, y_train)
    
    # Prédiction et évaluation du modèle
    y_pred = model.predict(X_test_scaled)
    
    # Calcul des erreurs
    mae_scores.append(mean_absolute_error(y_test, y_pred))
    mse_scores.append(mean_squared_error(y_test, y_pred))

# Afficher la moyenne des erreurs pour chaque split
print(f"MAE moyen: {np.mean(mae_scores)}")
print(f"MSE moyen: {np.mean(mse_scores)}")
```

```python
# Ajouter une moyenne mobile sur les prédictions pour visualiser la tendance
rolling_pred = pd.Series(y_pred, index=y_test.index).rolling(window=5).mean()

plt.figure(figsize=(12, 6))
plt.plot(y_test.index, y_test, label='Valeurs réelles', color='blue')
plt.plot(y_test.index, y_pred, label='Prédictions', color='red', linestyle='--')
plt.plot(rolling_pred, label='Moyenne mobile des prédictions (5 jours)', color='green', linestyle='-.')

plt.title('Série Temporelle avec Moyenne Mobile')
plt.xlabel('Date')
plt.ylabel('Quantité utilisée')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from keras.models import Sequential
from keras.layers import LSTM, Dense

# Exemple de série temporelle
df['Production_Start_Time'] = pd.to_datetime(df['Production_Start_Time'], format='%Y%m%d%H')

# Préparer la série temporelle
time_series = df['quantityused'].values
time_series = time_series.reshape(-1, 1)

# Normalisation des données (LSTM nécessite des valeurs entre 0 et 1)
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(time_series)

# Créer un jeu de données de séquences pour l'apprentissage
def create_dataset(data, time_step=1):
    X, y = [], []
    for i in range(len(data) - time_step - 1):
        X.append(data[i:(i + time_step), 0])
        y.append(data[i + time_step, 0])
    return np.array(X), np.array(y)

# Définir le nombre de pas de temps (ex. 24 pour 24 heures)
time_step = 24

X, y = create_dataset(scaled_data, time_step)

# Reshaper les données pour le LSTM (samples, time_steps, features)
X = X.reshape(X.shape[0], X.shape[1], 1)

# Créer le modèle LSTM
model = Sequential()
model.add(LSTM(units=50, return_sequences=True, input_shape=(X.shape[1], 1)))
model.add(LSTM(units=50, return_sequences=False))
model.add(Dense(units=1))

# Compiler et entraîner le modèle
model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(X, y, epochs=20, batch_size=32)

# Prédiction pour les 24 prochaines heures (par exemple)
predictions_24 = model.predict(X[-24:].reshape(1, 24, 1))

# Inverser la normalisation
predictions_24 = scaler.inverse_transform(predictions_24)

# Affichage des résultats pour les 24 prochaines heures
plt.figure(figsize=(10, 6))
plt.plot(range(24), predictions_24.flatten(), label='Prédictions 24 heures', color='red')
plt.title('Prédiction pour les 24 prochaines heures avec LSTM')
plt.xlabel('Heure')
plt.ylabel('Quantité utilisée')
plt.legend()
plt.grid(True)
plt.show()
```
